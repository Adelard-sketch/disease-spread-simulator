# Algorithm Implementation Details

# Chosen Algorithm: Random Forest Classifier

We selected the Random Forest Classifier for its robustness to overfitting, ability to handle imbalanced data, and feature importance insights, as informed by StatQuest tutorials (Bamford, 2020). Random Forest is an ensemble method that builds multiple decision trees on bootstrapped samples of the data, with the final prediction being the mode of individual tree predictions. Mathematically, for features (X), target (y), and (T) trees:

Random feature selection at each split reduces tree correlation, enhancing generalization.

## Implementation

We implemented the Random Forest using scikit-learn’s `RandomForestClassifier` (Pedregosa et al., 2011) with the following steps:

1. **Data Splitting**: Split the preprocessed dataset of 5,861,480 entries into 80% training and 20% test sets using `train_test_split`, with stratification to preserve class distribution. The test set contains 1,172,296 entries.
2. **Class Imbalance**: Applied SMOTE (Synthetic Minority Oversampling Technique) to the training set, generating synthetic samples for the minority class (positive cases) to address the 90% negative, 10% positive imbalance.
3. **Model Initialization**: Initialized the Random Forest model with a fixed random state for reproducibility:
   - `RandomForestClassifier(random_state=42)`
4. **Hyperparameter Tuning**: Used `GridSearchCV` with a 3-fold stratified cross-validation via `StratifiedKFold(n_splits=3, shuffle=True, random_state=42)` to tune the following parameters:
   - `n_estimators`: \[10, 10\]
   - `max_depth`: \[10, 20\]
   - `min_samples_split`: \[2, 5\]
   - `bootstrap`: \[True\] The smaller grid was chosen for faster execution in Google Colab. The best parameters were determined through this process, optimizing for performance on the imbalanced dataset.
5. **Training and Saving**: Trained the model on the balanced training set and saved it as `covid_prediction_model.pkl` using `joblib`.

## Model Performance

The Random Forest model’s performance on the test set (1,172,296 entries) is summarized below:

**Confusion Matrix**:

- True Negatives: 986,684
- False Positives: 84,477
- False Negatives: 40,636
- True Positives: 60,499

**Classification Report**:

- **Class 0 (Negative)**: Precision 0.96, Recall 0.92, F1-score 0.94, Support 1,071,161
- **Class 1 (Positive)**: Precision 0.42, Recall 0.60, F1-score 0.49, Support 101,135
- **Accuracy**: 0.89 (89%)
- **Macro Avg**: Precision 0.69, Recall 0.76, F1-score 0.72
- **Weighted Avg**: Precision 0.91, Recall 0.89, F1-score 0.90

Random Forest outperformed baseline models like Logistic Regression and Decision Trees, which achieved lower F1-scores on the minority class during initial testing. The model excels at identifying negative cases (high precision and recall for Class 0), but struggles with the minority class (Class 1), as indicated by the lower precision (0.42) despite a decent recall (0.60). Feature importance analysis revealed that `test_indication_Contact with confirmed` (0.3088) and `test_indication_Other` (0.2941) were the most predictive features, emphasizing the role of exposure history in infection risk prediction. Visualizations like confusion matrices and classification reports were generated using Matplotlib and integrated into our Streamlit app for interpretability.
